# Manipulasi Data Sederhana Dengan Statistik Deskriptif Dari Suatu Data Frame

**Statistik deskriptif**, atau yang dikenal sebagai summary dalam library Pandas dapat diperoleh dengan fungsi **.describe()**. Function ini dapat memberikan informasi mengenai nilai rataan, standar deviasi, serta *IQR (interquartile range)* dari suatu data.

Secara default, function **.describe()** secara otomatis akan mengabaikan kolom category dan hanya memberikan summary statistik untuk kolom berjenis numerik.

praktek berikut akan mengevaluasi performa cabang Toko A dengan quick summary dari segi kuantitas , harga , freight value , dan weight yang dibeli konsumen. Ada juga median dari total pembelian konsumen per transaksi yang harus kita sertakan di dalamnya untuk bisa dianalisis.

Supaya kita bisa mendapatkan summary statistik atau statistik deskriptif dari kolom numerik dan karakter, kita perlu menambahkan argumen yaitu **include = “all”**.

Untuk bisa memperoleh data quick summary tersebut, kita akan menggunakan fungsi**.describe()** terhadap order_df untuk melihat statistik deskriptif dari kolom *quantity, price, freight_value,* dan *product_weight_gram* yang masing-masing mewakili kuantitas, harga, freight value, serta weight dari produk yang dibeli konsumen.

Kemudian, kita akan melihat nilai median dari kolom price dengan menggunakan fungsi .median(). Dataframe yang akan kita gunakan dalam praktek ini adalah data frame order_df.

```python
#Jawaban Project Praktek
import pandas as pd

# Memuat dataset dari URL yang diberikan
url = "https://raw.githubusercontent.com/rahmadsa/dataset/main/data_order_cabang_a.csv"
order_df = pd.read_csv(url)

# Menampilkan ringkasan statistik deskriptif untuk SEMUA kolom (numerik dan objek/kategori)
# Menggunakan argumen include="all"
print("=== Ringkasan Statistik Deskriptif (Semua Kolom) ===")
print(order_df.describe(include="all"))
print("\n")

# Menghitung nilai Median khusus untuk kolom "price"
median_price = order_df.loc[:, "price"].median()
print(f"Median dari total pembelian konsumen (kolom price): {median_price}")
```

penjelasan :<br>
pd.read_csv(): Digunakan untuk membaca dataset berbentuk CSV secara daring langsung melalui repositori GitHub.

include="all": Secara bawaan (default), fungsi .describe() hanya memproses kolom yang bertipe numerik saja (seperti kuantitas dan harga). Dengan menambahkan parameter include="all", Pandas akan dipaksa untuk memunculkan statistik bagi kolom kategorikal/string. Untuk kolom non-numerik, ringkasan akan menampilkan metrik berupa count (jumlah data non-null), unique (jumlah nilai unik), top (nilai yang paling sering muncul), dan freq (frekuensi kemunculan nilai tertinggi tersebut).

.median(): Digunakan untuk memperoleh nilai tengah data pada kolom price setelah diurutkan, yang berfungsi memberikan gambaran nilai transaksi yang lebih stabil karena tidak terpengaruh oleh pencilan (outlier) atau nilai ekstrem yang terlalu tinggi/rendah.<br>
