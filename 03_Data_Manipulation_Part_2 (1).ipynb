## **4. Read Dataset - RDBMS (SQLite)**

Methode pembacaan RDBMS (SQLite) disajikan pada notebook berbeda

````python
import pandas as pd
import sqlite3

# 1. Membuat database SQLite sementara kmd diisi dengan data sampel
conn = sqlite3.connect(':memory:')
df_sample = pd.DataFrame({'id': [1, 2, 3], 'nama': ['Adi', 'Budi', 'Siti'], 'Nilai': [90, 85, 88]})
df_sample.to_sql('mahasiswa', conn, index=False)

# Membaca data dari RDBMS (SQLite) menggunakan Pandas
query = "SELECT * FROM mahasiswa WHERE Nilai >= 87"
df_db = pd.read_sql_query(query, conn)

# Tampilkan hasil
print("Hasil pembacaan dari RDBMS (SQLite):")
print(df_db)

conn.close()
````

penjelasan : Fungsi utama yang digunakan adalah pd.read_sql_query(query, connection). Fungsi ini menerima dua parameter wajib: perintah bahasa SQL (SELECT * FROM ...) dan objek koneksi database (conn).

Output yang dihasilkan otomatis dikonversi oleh Pandas menjadi struktur DataFrame yang siap dianalisis.<br>

## **5. Head Dan Tail**
Seperti yang telah dipelajari sebelumnya bahwa ada method .head yang diterapkan pada suatu variabel bertipe pandas dataframe/series.

Method .head ditujukan untuk membatasi tampilan jumlah baris teratas dari dataset. Sementara itu, method .tail ditujukan untuk membatasi jumlah baris terbawah dari dataset.

Secara umum kedua method ini memiliki bentuk


```python
[nama_dataframe].head(n)
```
dan

```python
[nama_dataframe].tail(n)
```

dengan **n** merupakan jumlah baris yang akan ditampilkan, jika tidak disebutkan **n = 5** (sebagai nilai default dari n).

![Kode Head & Tails](https://github.com/rahmadsa/Data-Wrangling-With-Python/blob/main/properties/img/2024-03-22-06-15-15.png?raw=true)

**Tugas Praktek 4 :**

Carilah refrensi cara menggabungkan beberapa dataframe kedalam sebuah dataframe baru.

Berikut disajikan dua buah dataset :

1. **Dataset csv** : https://github.com/rahmadsa/dataset/blob/main/sample_csv.csv

2. **Dataset tsv** : https://github.com/rahmadsa/dataset/blob/main/sample_tsv.tsv


*Requirement* :

1. Buatlah Dataset baru yang merupakan gabungan 2 (dua) dataset diatas dalam sebuah dataframe baru dengan nama  df_NIM_Anda contoh *df_19979922*.
2. Tampilkan jumlah masing-masing data pada dataframe point no 1 (dataset 1 dan dataset 2)
3. Tampilkan jumlah data pada dataset data baru dari requirement nomor 1
4. Tampilkan 10 data teratas dan 10 data terbawah, dari data frame baru point no 1







````python
#Jawaban
import pandas as pd

# Load dataset awal dari GitHub
url_csv = "https://raw.githubusercontent.com/rahmadsa/dataset/main/sample_csv.csv"
url_tsv = "https://raw.githubusercontent.com/rahmadsa/dataset/main/sample_tsv.tsv"

df_dataset1 = pd.read_csv(url_csv)
df_dataset2 = pd.read_csv(url_tsv, sep='\t')

df_12231945 = pd.concat([df_dataset1, df_dataset2], axis=0, ignore_index=True)

print("=== REQ 2: Jumlah Masing-masing Data Awal ===")
print("Jumlah baris Dataset 1 (CSV) :", len(df_dataset1), "baris")
print("Jumlah baris Dataset 2 (TSV) :", len(df_dataset2), "baris\n")

print("=== REQ 3: Jumlah Data pada Dataframe Baru ===")
print("Jumlah total baris df_12231945 :", len(df_12231945), "baris\n")

print("=== REQ 4: 10 Data Teratas (Head) ===")
print(df_12231945.head(10))
print("\n")

print("=== REQ 4: 10 Data Terbawah (Tail) ===")
print(df_12231945.tail(10))
````

Penjelasan : Untuk menggabungkan dua buah DataFrame yang memiliki struktur kolom yang sama, kita menggunakan fungsi pd.concat(). Parameter ignore_index=True ditambahkan agar indeks dari kedua tabel digabungkan secara berurutan kembali (mulai dari 0 hingga akhir), sehingga tidak terjadi duplikasi nomor indeks asal.

Berdasarkan hasil eksekusi, Dataset 1 memiliki 101 baris data dan Dataset 2 memiliki 101 baris data.

Setelah dilakukan penggabungan vertikal, jumlah total data baru di dalam tabel df_12231945 menjadi 202 baris data (101 + 101).

Fungsi .head(10) dan .tail(10) berhasil menampilkan 10 sampel baris awal dari data area Jakarta (Dataset 1) serta 10 sampel baris akhir dari data area luar Jakarta (Dataset 2).<br>
